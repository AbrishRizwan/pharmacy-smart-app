```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error
df=pd.read_csv("/content/salesdaily.csv")
df=df.drop(columns=["datum"])
df=pd.get_dummies(df,columns=["Weekday Name"],dtype=int)
medicine_cols=["M01AB","M01AE","N02BA","N02BE","N05B","N05C","R03","R06"]
x=df.drop(columns=medicine_cols)
models={}
mae_scores = {}
for med in medicine_cols:
    y=df[med]
    x_train, x_test, y_train, y_test =train_test_split(x,y,test_size=0.2,random_state=42)
    model = RandomForestRegressor(random_state=42)
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    mae = mean_absolute_error(y_test, y_pred)
    models[med] = model
    mae_scores[med] = mae
    print(f"Medicine: {med:<6} | Average Error (MAE): {mae:.2f} units")
joblib.dump(models, "medicine_models.pkl")
```
